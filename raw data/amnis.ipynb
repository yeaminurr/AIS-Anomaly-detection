```python
# start_time = SOME_START_TIME_VALUE  # Replace SOME_START_TIME_VALUE with the start time value
# end_time = SOME_END_TIME_VALUE  # Replace SOME_END_TIME_VALUE with the end time value


# Function to filter entries
def filter_by_track_time_range(entries, start_time, end_time):
    coordinate_data = []
    for entry in entries:
        filtered_entries = []
        # Check each track in the entry
        for track in entry['tracks']:
            track_time = track["time"]
            
            if start_time <= track_time <= end_time:
                #print(track_time)
                coordinates = track["location"]["coordinates"][::-1]
                #print("")
                filtered_entries.append(coordinates)
                #break  # Break after finding the first matching track to avoid duplicates
        #print(filtered_entries);
        if len(filtered_entries)>0:
            coordinate_data.append({"mmsi":entry["mmsi"],"coordinates":filtered_entries})
            
    return coordinate_data

# Apply the filter function
```
